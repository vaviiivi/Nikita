Цикл for: перебор списка с помощью индексов

```python
# Вывести индексы четных чисел
l = [1, 12, 5, 9, 10, -100, 12]
for x in l:
    if x % 2 == 0:
        print(l.index(x))  # Функция index выводит индекс первого найденного x
# Не подходит
```

```python
l = [1, 12, 5, 9, 10, -100, 12]
for index in range(0, len(l)):
    # print(l[index], index) число и его индекс
    if l[index] % 2 == 0:
        print(index)
```

В списке l заменить все отрицательные числа на нули.

```python
l = [1, 12, 5, -33, 9, 10, -100, 12, -5]
for index in range(0, len(l)):
    if l[index] < 0:
        l[index] = 0
print(l)
```

В списке l найти сумму индексов положительных чисел.

```python
l = [1, 12, 5, -33, 9, 10, -100, 12, -5]
summ=0
for index in range(0,len(l)):
    if l[index]>0:
        summ=summ+index
print(summ)
```

В списке l найти максимальное число и его индекс.

```python
l = [1, 12, 5, -33, 9, 10, -100, 12, -5]

maxx = max(l)  # 12

for index in range(0, len(l)):
    if l[index] == maxx:
        print(maxx, index)
        break
```

В списке l найти максимальное и минимальное число и их индексы.

```python
l = [1, 12, 5, -33, 9, 10, -100, -5]

maxx = max(l)  # 12
minn = min(l)  # -100

for index in range(0, len(l)):
    if l[index] == maxx:
        print(maxx, index)
    if l[index] == minn:
        print(minn, index)
```

В списке l найти максимальное и минимальное число и поменять их местами.

```python
l = [1, 12, 5, -33, 9, 10, -100, -5]

maxx = max(l)  # 12
minn = min(l)  # -100

for index in range(0, len(l)):
    if l[index] == maxx:
        l[index] = minn
    elif l[index] == minn:
        l[index] = maxx

print(l)
```
